# Function to prepare the data that returns adjusted closing price DataFrame

```python
def data_prep(data):

    data = data.reset_index()
    df = data[["Date", "Adj Close", "Close", "High", "Low", "Open"]]
    adj_close_price_df = df["Adj Close"]
    adj_close_price_df.insert(loc = 0, column = "Date", value = df["Date"])
    
    return adj_close_price_df
```

```python
df = data_prep(data)
display(df.head())
```

# Creating a function to return Percentage Daily Returns DataFrame

```python
def percentage_daily_returns(adj_close_df):
    daily_returns_df = adj_close_df.copy()
    for column in daily_returns_df.columns[1:]:
        daily_returns_df[column] = adj_close_df[column].pct_change(1)
    
    daily_returns_df.replace(np.nan, 0, inplace = True)
    return daily_returns_df
```

```python
pdr_df = percentage_daily_returns(df)
display(pdr_df)
```

# Creating a Function to plot graphs

```python
def plot_financial_data(df, title):
    
    fig = px.line(title = title)
    
    for column in df.columns[1:]:

        fig.add_scatter(
            x = df["Date"],
            y = df[column],
            name = column
        )
        fig.update_layout(
            template = 'plotly_dark',
            xaxis_title = "Date"
        )
    fig.show()
```

```python
plot_financial_data(df, "Adjusted Closing Price of Stocks [$]")
```
